**Unsupervised Learning**

정답 없이 데이터의 유사성이나 숨은 구조를 찾는 방법.

소재 분야에서 이해하기: 비슷한 스펙트럼을 가진 시료를 묶는다.

이 노트북은 개념을 직접 돌려보기 위한 예제입니다. 데이터는 실제 측정값이 아니라 개념 확인용으로
생성한 값이므로, 결과 수치를 연구 결론으로 쓰지 마세요. 위에서부터 순서대로 실행하세요.
그림의 축 이름은 기본 폰트에 한글 글리프가 없어 영문으로 적었습니다.

## 1. 정답 없이 구조 찾기

세 종류의 공정에서 나온 시료를 섞어놓고, 라벨 없이 군집을 찾아봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

centers = np.array([[700, 1.0], [780, 5.0], [860, 2.5]])
groups = rng.integers(0, 3, 210)
data = centers[groups] + rng.normal(0, [18, 0.5], (210, 2))
plt.scatter(data[:, 0], data[:, 1], s=16, c='gray')
plt.xlabel('sintering temperature (C)'); plt.ylabel('hold time (h)'); plt.title('unlabelled samples'); plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

scaled = StandardScaler().fit_transform(data)
labels = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(scaled)
plt.scatter(data[:, 0], data[:, 1], c=labels, s=16, cmap='viridis')
plt.xlabel('sintering temperature (C)'); plt.ylabel('hold time (h)'); plt.title('discovered clusters'); plt.show()

from sklearn.metrics import adjusted_rand_score
print('실제 공정 구분과의 일치도(ARI) %.2f' % adjusted_rand_score(groups, labels))
```

## 2. 해석

군집화는 정답을 쓰지 않았지만 실제 공정 구분과 상당히 겹칩니다. 다만 군집 번호는 임의이고,
군집 수를 몇 개로 볼지는 사람이 정해야 합니다. 스케일을 맞추지 않으면 단위가 큰 변수가 결과를 지배합니다.

```python
# 스케일을 맞추지 않으면 온도(수백 단위)가 시간(한 자리)을 압도합니다.
raw_labels = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(data)
print('스케일링 없이 ARI %.2f' % adjusted_rand_score(groups, raw_labels))
```
