## 1. 층을 쌓아 비선형 관계 학습

작은 신경망으로 곡선 관계를 학습시켜 봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

x = np.linspace(0, 1, 200)
truth = np.sin(3 * np.pi * x) * np.exp(-x)
obs = truth + rng.normal(0, 0.06, x.size)
plt.plot(x, truth, 'k--', label='true function'); plt.scatter(x, obs, s=10, label='observations')
plt.legend(); plt.show()
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.linear_model import LinearRegression

deep = MLPRegressor(hidden_layer_sizes=(64, 64), max_iter=8000, random_state=0).fit(x[:, None], obs)
linear = LinearRegression().fit(x[:, None], obs)
plt.plot(x, truth, 'k--', label='true function')
plt.plot(x, deep.predict(x[:, None]), label='MLP (64,64)')
plt.plot(x, linear.predict(x[:, None]), label='linear regression')
plt.legend(); plt.show()
```

## 2. 층 수를 바꿔보기

```python
for layers in [(4,), (16,), (64, 64), (128, 128, 128)]:
    model = MLPRegressor(hidden_layer_sizes=layers, max_iter=8000, random_state=0).fit(x[:, None], obs)
    error = np.mean(np.abs(model.predict(x[:, None]) - truth))
    print('%-18s 참 함수와의 평균 오차 %.4f' % (str(layers), error))
```

## 3. 해석

층과 폭이 커지면 표현력이 늘지만, 데이터가 적으면 잡음까지 학습해 오히려 나빠집니다.
딥러닝의 이점은 데이터가 충분하고 관계가 복잡할 때 나타납니다.
