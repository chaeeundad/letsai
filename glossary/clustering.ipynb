## 1. 군집 수를 정하는 문제

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

centres = np.array([[0, 0], [4, 4], [8, 0], [4, -4]])
groups = rng.integers(0, 4, 400)
data = centres[groups] + rng.normal(0, 0.9, (400, 2))
plt.scatter(data[:, 0], data[:, 1], s=12, c='gray'); plt.title('unlabelled data'); plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 8):
    labels = KMeans(k, n_init=10, random_state=0).fit_predict(data)
    print('k=%d  관성 %8.1f  실루엣 %.3f' % (k, KMeans(k, n_init=10, random_state=0).fit(data).inertia_,
                                          silhouette_score(data, labels)))
```

```python
from sklearn.cluster import DBSCAN, AgglomerativeClustering

moon = np.vstack([np.column_stack([np.linspace(0, 3, 200), np.sin(np.linspace(0, 3, 200))]),
                  np.column_stack([np.linspace(0, 3, 200), np.sin(np.linspace(0, 3, 200)) + 1.2])])
moon = moon + rng.normal(0, 0.05, moon.shape)
fig, axes = plt.subplots(1, 3, figsize=(12, 3.2))
for axis, (name, model) in zip(axes, [('k-means', KMeans(2, n_init=10, random_state=0)),
                                      ('agglomerative', AgglomerativeClustering(2)),
                                      ('DBSCAN', DBSCAN(eps=0.25, min_samples=5))]):
    axis.scatter(moon[:, 0], moon[:, 1], c=model.fit_predict(moon), cmap='coolwarm', s=8)
    axis.set_title(name)
plt.tight_layout(); plt.show()
print('k-means는 구형 군집을 가정하므로 띠 모양 구조에서는 원하는 결과를 주지 못합니다.')
```

또 알고리즘마다 가정하는 군집 모양이 달라, 데이터 구조에 맞는 방법을 골라야 합니다.
