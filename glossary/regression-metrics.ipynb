# 회귀 성능 지표 실습

**MAE · RMSE · R²**

예측 오차의 크기와 설명력을 나타내는 지표. 단위와 이상치 민감도가 서로 다르다.

소재 분야에서 이해하기: RMSE는 큰 오차에 더 민감하므로 MAE와 함께 본다.

이 노트북은 개념을 직접 돌려보기 위한 예제입니다. 데이터는 실제 측정값이 아니라 개념 확인용으로
생성한 값이므로, 결과 수치를 연구 결론으로 쓰지 마세요. 위에서부터 순서대로 실행하세요.
그림의 축 이름은 기본 폰트에 한글 글리프가 없어 영문으로 적었습니다.

참고 자료: [scikit-learn 평가 지표 문서](https://scikit-learn.org/stable/modules/model_evaluation.html)

## 1. 지표마다 보는 것이 다릅니다

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

truth = rng.uniform(100, 300, 500)
small_errors = truth + rng.normal(0, 10, 500)                       # 고르게 조금씩 틀림
mostly_good = truth + rng.normal(0, 3, 500)
mostly_good[rng.choice(500, 10, replace=False)] += 120              # 대부분 정확하지만 큰 실수 10건
print('두 예측을 비교합니다.')
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, median_absolute_error

for name, prediction in [('고르게 조금씩 틀림', small_errors), ('대부분 정확 + 큰 실수 10건', mostly_good)]:
    print('%-24s MAE %6.2f  RMSE %6.2f  중위절대오차 %5.2f  R2 %.3f'
          % (name, mean_absolute_error(truth, prediction),
             np.sqrt(mean_squared_error(truth, prediction)),
             median_absolute_error(truth, prediction), r2_score(truth, prediction)))
print('\nRMSE는 큰 오차를 제곱해 강조하므로 두 번째 예측을 훨씬 나쁘게 평가합니다.')
```

## 2. R²는 대상의 분산에 따라 달라집니다

```python
for spread in (5, 20, 100):
    narrow_truth = rng.normal(200, spread, 500)
    prediction = narrow_truth + rng.normal(0, 10, 500)
    print('목표값 표준편차 %3d -> MAE %.2f (동일) / R2 %.3f'
          % (spread, mean_absolute_error(narrow_truth, prediction), r2_score(narrow_truth, prediction)))
print('\n같은 절대 오차라도 목표값의 분산이 작으면 R2 는 낮게 나옵니다.')
print('논문 사이에서 R2 만 비교하면 오해할 수 있어, 단위가 있는 MAE/RMSE를 함께 보고해야 합니다.')
```
