## 1. 층과 활성함수

은닉층 없이는 선형 경계만 만들 수 있습니다. XOR 형태의 문제로 확인합니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

n = 600
points = rng.uniform(-1, 1, (n, 2))
labels = ((points[:, 0] * points[:, 1]) > 0).astype(int)
plt.scatter(points[:, 0], points[:, 1], c=labels, cmap='coolwarm', s=12)
plt.xlabel('feature 1'); plt.ylabel('feature 2'); plt.title('XOR-like problem'); plt.show()
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

for name, model in [('로지스틱(층 없음)', LogisticRegression()),
                    ('MLP (4,)', MLPClassifier((4,), max_iter=4000, random_state=0)),
                    ('MLP (32,32)', MLPClassifier((32, 32), max_iter=4000, random_state=0))]:
    print('%-18s 정확도 %.3f' % (name, cross_val_score(model, points, labels, cv=5).mean()))
```

```python
model = MLPClassifier((32, 32), max_iter=4000, random_state=0).fit(points, labels)
gx, gy = np.meshgrid(np.linspace(-1, 1, 200), np.linspace(-1, 1, 200))
zone = model.predict(np.column_stack([gx.ravel(), gy.ravel()])).reshape(gx.shape)
plt.contourf(gx, gy, zone, alpha=0.3, cmap='coolwarm')
plt.scatter(points[:, 0], points[:, 1], c=labels, cmap='coolwarm', s=10)
plt.title('decision boundary of the MLP'); plt.show()
```

## 2. 해석

은닉층과 비선형 활성함수가 있어야 곡선 경계를 만들 수 있습니다. 층을 늘리면 표현력은 오르지만
데이터가 부족하면 과적합합니다. 표 형태 데이터에서는 트리 앙상블이 더 나은 경우가 많습니다.
