# 과적합 실습

**Overfitting**

학습 데이터에는 잘 맞지만 새로운 데이터에서는 성능이 떨어지는 상태.

소재 분야에서 이해하기: 기존 합금만 잘 맞추고 새 조성에서는 오차가 커진다.

이 노트북은 개념을 직접 돌려보기 위한 예제입니다. 데이터는 실제 측정값이 아니라 개념 확인용으로
생성한 값이므로, 결과 수치를 연구 결론으로 쓰지 마세요. 위에서부터 순서대로 실행하세요.
그림의 축 이름은 기본 폰트에 한글 글리프가 없어 영문으로 적었습니다.

참고 자료: [scikit-learn 용어집](https://scikit-learn.org/stable/glossary.html)

## 1. 학습 오차와 시험 오차의 분리

다항식 차수를 올려가며 두 오차를 함께 봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

x = np.sort(rng.uniform(0, 1, 30))
truth = np.sin(2 * np.pi * x)
observed = truth + rng.normal(0, 0.2, x.size)
x_test = np.sort(rng.uniform(0, 1, 200))
y_test = np.sin(2 * np.pi * x_test) + rng.normal(0, 0.2, x_test.size)
plt.scatter(x, observed, label='training data'); plt.plot(x_test, np.sin(2 * np.pi * x_test), 'k--', label='true')
plt.legend(); plt.show()
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

degrees, train_errors, test_errors = range(1, 16), [], []
for degree in degrees:
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression()).fit(x[:, None], observed)
    train_errors.append(mean_squared_error(observed, model.predict(x[:, None])))
    test_errors.append(mean_squared_error(y_test, model.predict(x_test[:, None])))
    print('차수 %2d  학습 MSE %.4f  시험 MSE %.4f' % (degree, train_errors[-1], test_errors[-1]))

plt.plot(list(degrees), train_errors, 'o-', label='training')
plt.plot(list(degrees), test_errors, 's-', label='test')
plt.yscale('log'); plt.xlabel('polynomial degree'); plt.ylabel('MSE'); plt.legend(); plt.show()
```

## 2. 해석

학습 오차는 차수를 올릴수록 계속 줄지만 시험 오차는 어느 지점부터 다시 커집니다.
그 지점을 넘어선 모델은 잡음을 외운 것입니다. 학습 오차만 보고 모델을 고르면 안 되는 이유입니다.
