## 1. 분포를 학습해 새 후보 만들기

기존 조성 데이터의 분포를 가우시안 혼합으로 학습하고, 그 분포에서 새 후보를 뽑아봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

known = np.vstack([rng.normal([0.6, 0.3], [0.05, 0.04], (120, 2)),
                   rng.normal([0.3, 0.6], [0.04, 0.05], (120, 2))])
known = np.clip(known, 0.01, 0.98)
plt.scatter(known[:, 0], known[:, 1], s=14, label='known compositions')
plt.xlabel('element A fraction'); plt.ylabel('element B fraction'); plt.legend(); plt.show()
```

```python
from sklearn.mixture import GaussianMixture

generator = GaussianMixture(n_components=2, random_state=0).fit(known)
new, _ = generator.sample(80)
plt.scatter(known[:, 0], known[:, 1], s=14, label='known')
plt.scatter(new[:, 0], new[:, 1], s=18, marker='x', label='generated')
plt.xlabel('element A fraction'); plt.ylabel('element B fraction'); plt.legend(); plt.show()
print('생성 후보 중 분율 합이 1을 넘는 물리적으로 불가능한 후보: %d개'
      % int((new.sum(axis=1) > 1).sum()))
```

## 2. 해석

생성 모델은 학습한 분포를 닮은 후보를 만들지만, 제약을 모르면 불가능한 후보도 만듭니다.
생성 결과는 반드시 물리적 타당성과 계산·실험으로 검증해야 합니다.
