## 1. 문장을 벡터로

외부 모델 없이 TF-IDF로 초록 문장을 벡터화하고 유사한 문헌을 찾아봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
plt.rcParams['figure.figsize'] = (7, 4)

docs = [
 'graph neural network predicts formation energy of oxides',
 'machine learning interatomic potential for molecular dynamics of alloys',
 'bayesian optimization guides synthesis temperature of thin films',
 'neural network potential accelerates molecular dynamics simulation',
 'x-ray diffraction analysis of sintered ceramic microstructure',
 'dilatometry measurement of thermal expansion in ceramics',
]
for i, d in enumerate(docs):
    print(i, d)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectors = TfidfVectorizer().fit_transform(docs)
similarity = cosine_similarity(vectors)
print('문헌 1과 가장 비슷한 문헌:', int(np.argsort(-similarity[1])[1]))
plt.imshow(similarity, cmap='viridis'); plt.colorbar(label='cosine similarity')
plt.xlabel('document'); plt.ylabel('document'); plt.show()
```

## 2. 해석

벡터 사이 거리가 가까우면 내용이 비슷하다고 봅니다. TF-IDF는 단어가 겹칠 때만 비슷하다고
판단하므로, 표현이 다르고 뜻이 같은 문장은 놓칩니다. 의미까지 담으려면 학습된 임베딩 모델이 필요합니다.
